# Vendi score: how many ideas is this list worth?

The Vendi score (Friedman & Dieng, 2023, [arXiv:2210.02410](https://arxiv.org/abs/2210.02410)) answers one question: *given how alike these items are, how many truly distinct items are they equivalent to?*

It is also prep for the Apart epistemics sprint: parts 2 and 4 check whether embedding similarity can be trusted, which the sprint's metric depends on. ## Part 1: the grid, by hand

**The contract.**

- **Input:** an n×n similarity grid K. It is symmetric, has 1s on the diagonal, and has entries between 0 (unrelated) and 1 (the same idea).
- **Output:** a number between 1 and n, the effective number of distinct items.
- **Recipe, in words:** divide K by n, take its eigenvalues, compute their Shannon entropy, then exponentiate.
